```python
import cv2
from skimage.metrics import structural_similarity as ssim 

a=cv2.imread("images3.jpg",cv2.IMREAD_GRAYSCALE) 
b=cv2.imread("images3.jpg",cv2.IMREAD_GRAYSCALE)

b=cv2.resize(b,(a.shape[1],a.shape[0]))

score=ssim(a,b)

print(f"Similarity:{score:.1%}")

```
